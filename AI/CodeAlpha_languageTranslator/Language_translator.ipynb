### Swahili to English translation

```python
#importing libraries
from tkinter import *
from tkinter import ttk
from googletrans import Translator,LANGUAGES
```

```python
root = Tk()
root.geometry('1200x500')
root.resizable(0,0)
root['bg'] = '#f0f0f0'

root.title("Language Translator")

# Title
title_label = Label(root, text='Language Translator', font=('Arial', 24, 'bold'), bg='#f0f0f0', fg='#1a1a1a')
title_label.pack(pady=20)

# Subtitle
subtitle = Label(root, text='Translate between languages instantly', font=('Arial', 10), bg='#f0f0f0', fg='#666666')
subtitle.pack()

# Input Label
input_label = Label(root, text='📝 Enter your text', bg='#f0f0f0', font=('Arial', 12, 'bold'), fg='#1a1a1a')
input_label.place(x=50, y=90)

# Output Label
output_label = Label(root, text='✓ Translation', bg='#f0f0f0', font=('Arial', 12, 'bold'), fg='#1a1a1a')
output_label.place(x=650, y=90)
```

```python
input_text = Entry(root, width=55, font=('Arial', 11), borderwidth=2, relief='solid', bg='white')
input_text.place(x=50, y=130)
input_text.config(highlightthickness=1, highlightcolor='#4CAF50')
```

```python
output_text = Text(root, font=('Arial', 11), height=8, wrap=WORD, padx=10, pady=10, width=48, borderwidth=2, relief='solid', bg='#fafafa')
output_text.place(x=650, y=130)

language = list(LANGUAGES.values())

# Language selection frame
lang_frame = Frame(root, bg='#f0f0f0')
lang_frame.place(x=50, y=240)
Label(lang_frame, text='Select Language:', font=('Arial', 11, 'bold'), bg='#f0f0f0', fg='#1a1a1a').pack(side=LEFT, padx=(0,10))

dest_lang = ttk.Combobox(lang_frame, values=language, width=35, font=('Arial', 10), state='readonly')
dest_lang.pack(side=LEFT)
dest_lang.set('english')

def Translate():
    try:
        if not input_text.get().strip():
            output_text.delete(1.0, END)
            output_text.insert(END, 'Please enter text to translate.')
            return
        
        lang_code = [k for k, v in LANGUAGES.items() if v == dest_lang.get()]
        if not lang_code:
            output_text.delete(1.0, END)
            output_text.insert(END, 'Please select a valid language.')
            return
        
        translator = Translator()
        translated = translator.translate(text=input_text.get(), dest=lang_code[0])
        output_text.delete(1.0, END)
        output_text.insert(END, translated.text)
    except Exception as e:
        output_text.delete(1.0, END)
        output_text.insert(END, f'Error: {str(e)}')

# Translate button with improved styling
trans_btn = Button(root, text='🔄 Translate', font=('Arial', 12, 'bold'), pady=10, padx=20, command=Translate, bg='#4CAF50', fg='white', activebackground='#45a049', relief='flat', borderwidth=0, cursor='hand2')
trans_btn.place(x=300, y=380)

# Clear button
def Clear():
    input_text.delete(0, END)
    output_text.delete(1.0, END)

clear_btn = Button(root, text='🗑️ Clear', font=('Arial', 12, 'bold'), pady=10, padx=20, command=Clear, bg='#f44336', fg='white', activebackground='#da190b', relief='flat', borderwidth=0, cursor='hand2')
clear_btn.place(x=550, y=380)

root.mainloop()
```
